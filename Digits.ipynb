```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_digits

# fix seed for reproducibility
np.random.seed(42)
```

```python
train_digits = []
with open('/ZipDigits.train') as fp:
    for line in fp:
        train_digits.append([float(x) for x in line.split()])

digits = np.array(train_digits)
```

```python
# from google.colab import drive
# drive.mount('/content/drive')
```

```python
digits_target = digits[:, -1]
```

```python
# Show 10 sample images — one per digit class — in a 2x5 grid
fig, axes = plt.subplots(2, 5, figsize=(10, 5))

local_digits_target = digits[:, 0]
local_digits_features = digits[:, 1:]

# Assuming ZipDigits.train images are 16x16, so 256 features.
image_shape = (16, 16)

for digit in range(10):
    # Find the first image of this digit class
    indices = np.where(local_digits_target == digit)[0]

    if len(indices) > 0:
        idx = indices[0]
        ax = axes[digit // 5][digit % 5]
        # Reshape the feature vector into an image
        ax.imshow(local_digits_features[idx].reshape(image_shape), cmap='gray_r')
        ax.set_title(f"Digit: {digit}", fontsize=12)
        ax.axis('off')
    else:
        # If no samples for a digit are found, add a placeholder
        ax = axes[digit // 5][digit % 5]
        ax.set_title(f"Digit: {digit}\n(No samples)", fontsize=10, color='red')
        ax.axis('off')

plt.suptitle("One sample image per digit class (16×16, values -1–1)", fontsize=13, y=1.02) # Adjusted title for ZipDigits
plt.tight_layout()
plt.show()
```
